```python
import re
```

```python
text = "I love python porgramming"
match = re.search(r'python', text)
print(match.group())
```

```python
text = "hello world"
match = re.search(r'^hello', text)
print(match.group())
```

```python
text = "hello world"
match = re.search(r'world$', text)
print(match.group())
```

```python
# vowles
text = "in my name xyz , ABC"
match = re.findall(r'[aeiou]', text)
print(match)
```

```python
# not vowel
text = "in my name xyz , ABC"
match = re.findall(r'[^aeiou]', text)
print(match)
```

```python
text = '1234 and 56'
match = re.findall(r'\d', text)
print(match)
```

```python
text = '1234 and 56'
match = re.findall(r'\d+', text)
print(match)
```

```python
text ='i like the red color and blue colour'

match= re.findall(r'colou?r', text)
print(match)
```

```python
# grouping and capturing

text = "Call me at 123-456-7890"
match = re.search(r'(\d{3})-(\d{3})-(\d{4})', text)
print(match.group())
```

```python
text= "1234 and 5678"
match=re.search(r'(?:\d{4})', text)

match.group()
```

```python
text = 'abc231'    #only followed by 123
match =re.search(r"abc(?=231)",text)
match.group()
```

```python
text = "i love programming"

match = re.search(r"\bprogramming\b", text)
print(match.group())
```

```python
text = "my phone number is 1234"

mod= re.sub(r'\d+', '4321', text)
print(mod)
```
