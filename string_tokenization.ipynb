```python
# String tokenization --using nltk, spacy

import nltk

from nltk.tokenize import word_tokenize, sent_tokenize

nltk.download("punkt_tab")
```

```python
text = "Hello World! NLP is amazing, Let's Tokenize this sentence"

print("word tokenize: ", len(word_tokenize(text)))
print("word tokenize: ", word_tokenize(text))
print("sentence tokenize: ", sent_tokenize(text))
print("sentence tokenize: ", len(sent_tokenize(text)))
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

print([token.text for token in doc])
print(len([sent.text for sent in doc.sents]))
```

```python
# basic word count and freq distribution

from collections import Counter
words = word_tokenize(text.lower())
word_freq = Counter(words)

print(len(words))
print(word_freq)
```

```python
# stopwords removal and frequency analysis

from nltk.corpus import stopwords
nltk.download("stopwords")
import contractions

stop_words = set(stopwords.words('english'))
text = contractions.fix(text)
doc = nlp(text)
words = [token.text for token in doc if not token.is_punct]
filtered_words = [word for word in words if word not in stop_words]
filtered_freq = Counter(filtered_words)

print(filtered_freq)
```

```python
# NER with context analysis

import spacy
from collections import Counter

text = "Apple is looking at buying a UK startup for 1 billion USD."

doc = nlp(text)

entity_freq = Counter([ent.text for ent in doc.ents])
for ent, freq in entity_freq.most_common():
  print(f"Entity: {ent}, Count: {freq}, Label: {nlp(ent)[0].ent_type_}")
```

```python
# N_Gram with frequency analysis
from nltk.util import ngrams
bigrams = list(ngrams(words, 2))
trigrams = list(ngrams(words, 3))

bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)

print(bigram_freq.most_common(5))
print(trigram_freq.most_common(5))
```

```python
# Keywords extraction using TF-IDF with Tokenization
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X= vectorizer.fit_transform([text])

feature_array= vectorizer.get_feature_names_out()

importance = X.toarray().flatten()
importance_words = sorted(zip(feature_array, importance), key=lambda x:x[1], reverse =True)[:5]

print([word for word, _ in importance_words])
```

```python
# text summarization
text = """
This is a string tokenization lecture. We are on a way to learn NLP
effectively. we hope we can excel in our career after learning NLP.
"""

doc = nlp(text)
word = [token.text for token in doc if token.is_alpha  and token.text not in stopwords.words("english")]
filtered_freq =Counter(word)
sentence_score ={}

for sent in doc.sents:
  for word in word_tokenize(sent.text.lower()):
    if word in filtered_freq:
      sentence_score[sent.text] = sentence_score.get(sent.text, 0)+ filtered_freq[word]

sorted_sentence = sorted(sentence_score, key = sentence_score.get, reverse =True)
summary = " ".join(sorted_sentence[:2])
print(summary)
```
