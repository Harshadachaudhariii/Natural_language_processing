```python
import re

def count_words(text):
  words = re.findall(r'\b\w+\b', text)

  return len(words)

text = "hello world !   , welcome to nlp."
count_words(text)
```

```python
from nltk import word_tokenize

def count_words_nltk(text):
  words = word_tokenize(text)
  return len(words)


text = "Let's Explore NLP with Python"
print(word_tokenize(text))
print("Word Count: ", count_words_nltk(text))
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
def count_words_spacy(text):
  docs = nlp(text)
  words = [token.text  for token in docs if not token.is_punct ]
  return len(words)

text = "Let's Explore NLP with Python"
print("Word count: ", count_words_spacy(text))
```

```python
from collections import Counter

def word_frequency(text):
  words = text.lower().split()
  return Counter(words)

text ="NLP is fun, NLP is powerful, and NLP is amazing in the future."
print("Word Frequency: ", word_frequency(text))
```

```python
# nlp -task 1 large sentence in small sentence

from nltk.tokenize import sent_tokenize, word_tokenize

def categorize_sentence(text):
  sentences = sent_tokenize(text)
  categorized = {"short": [], "long":[]}

  for sentence in sentences:
    docs = nlp(sentence)
    word_count = len([token  for token in docs if not token.is_punct ])
    if word_count < 5:
      categorized['short'].append(sentence)
    else:
      categorized['long'].append(sentence)

  return categorized


text = "Hello, NLP is fun. i like it very much, I hope everyone is following this lecture."
category = categorize_sentence(text)

print(category)
print("Short sentence: ", category['short'])
print("long sentence: ", category['long'])
```

```python
def question_categorize(text):
  sentences = sent_tokenize(text)
  categorized = {"Questions": [], "Not a questions":[]}

  for sentence in sentences:
    if sentence.strip().endswith("?"):
      categorized['Questions'].append(sentence)
    else:
      categorized['Not a questions'].append(sentence)

  return categorized

text = "what is nlp?"
category = question_categorize(text)

print(category)
```

```python
# Rule based sentiment analysis

import time
positive_word ={"happy", "good", "great", "excellent", "love", "amazing"}
negative_word ={"sad", "bad", "worse","terrible", "hate", "awful", "worst"}

def sentiment_analysis(text):
  start_time = time.time()

  word = text.lower().split()

  pos_count = sum(1 for word in word if word in positive_word)
  neg_count = sum(1 for word in word if word in negative_word)

  sentimet = "Positve sentiment " if pos_count > neg_count else "Negative sentiment" if neg_count > pos_count else "Netural sentiment"

  end_time = time.time()
  execution_time =round( end_time - start_time)

  return sentimet, execution_time

text ="I am in a  happy mood, but i feel sad about my friend who had a terrible fall from the bike"

sentiment, execution_time = sentiment_analysis(text)
print(sentiment, execution_time)
```

```python
# Rule based sentiment analysis

import time
positive_word ={"win", "prize"}
negative_word ={"call", "feedback"}

def sentiment_analysis(text):
  start_time = time.time()

  word = text.lower().split()

  pos_count = sum(1 for word in word if word in positive_word)
  neg_count = sum(1 for word in word if word in negative_word)

  sentimet = "Spam " if pos_count > neg_count else "Not a spam" if neg_count > pos_count else "Netural sentiment"

  end_time = time.time()
  execution_time =round( end_time - start_time)

  return sentimet, execution_time

text ="Win big in the met gala fanclub"

sentiment, execution_time = sentiment_analysis(text)
print(sentiment, execution_time)
```
