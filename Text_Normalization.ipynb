```python
# special character

import string
text = "Hello world! Let's code: #python #AI."
translator =str.maketrans("", "", string.punctuation)
cleaned_text = text.translate(translator)

print(cleaned_text)
```

```python
# removing space
text = "      Hello              world"

cleaned_text = ' '.join(text.split())
print(cleaned_text)
```

```python
# handling  numbers

text = "I have 2 apples and 3 oranges"

import re
cleaned_text = re.sub(r'\d+', "", text)
print(cleaned_text)
```

```python
num_of_digits={
    '2':'two',
    '3': 'three'
}
```

```python
cleaned_text =""
for char in text:
  if char.isdigit():
    cleaned_text += num_of_digits[char]
  else:
    cleaned_text += char
print(cleaned_text)
```

```python

text = "visit http://google.com for more info. Contact us at contact@example.com #AI"

processed_text = re.sub(r'http\S+|www\S|https\S+', "URL ", text)
processed_text = re.sub(r'\S+@\S+', "EMAIL ", processed_text)
processed_text = re.sub(r'#(\S+)', r'HASHTAG \1', processed_text)
processed_text
```

```python
from spellchecker import SpellChecker

text = 'I recieved a color or colour message ganrantee'
spell = SpellChecker()
words = text.split()
corrected_text = " ".join([spell.correction(word) for word in words])
print(corrected_text)
```
