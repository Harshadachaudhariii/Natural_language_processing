##Tokenization


```python
import nltk
# nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize

text_word = "I love programming"
sent_word = "i love programming, and i also love python. It is my passion"

token_word = word_tokenize(text_word)
print(token_word)
```

```python
token_sent = sent_tokenize(sent_word)
print(token_sent)
```

```python
for i in token_sent:
  x = word_tokenize(i)
  print(x)
```

##Stopword Removal

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
stop_words= set(stopwords.words("english"))
len(stop_words)
```

```python
text = "i haven't thought this through by doing most of these against themselves"
len(text)
```

```python
token = word_tokenize(text)
print(token)
print(len(token))
```

```python
stop_words
```

```python
filtered_tokens= [word for word in token if word.lower() not in stop_words]
print(filtered_tokens)
```

##Remove Punctuation


```python
import string
text = """
i have a news, and it is great. I am going to share it, but only with those people
who have helped me. And especially those, you have always criticised my work.
"""
```

```python
translator = str.maketrans("","", string.punctuation)
```

```python
processed = text.translate(translator)
processed
```

##stemming

```python
from nltk.stem import PorterStemmer
words = ['runner', 'burnt','happily', 'easily']

stemmer = PorterStemmer()
stemmed = [stemmer.stem(word) for word in words]
stemmed
```

##Lemmatization

```python
import spacy
```

```python
nlp= spacy.load('en_core_web_sm')
```

```python
word = 'running, burnt,flying'
doc = nlp(word)
```

```python
lemmatized = [token.lemma_ for token in doc]
lemmatized
```

```python
for token in doc:
  print(f"{token.text: <15} {token.pos_:<10}{token.lemma_}")
```

##N_GRAM


```python
from nltk.util import ngrams

sample_text = "My name is John, i have 10 sec record in running 100 meters"
sample_text = sample_text.lower()

processed = sample_text.translate(translator)

tokenize = word_tokenize(processed)
sample_text = [word for word in tokenize if word not in stop_words]

bigram = list(ngrams(sample_text, 2))
bigram
```
